```python
import pandas as pd
```

```python

path = '/content/drive/MyDrive/e-commerce/'

orders = pd.read_csv(path + 'olist_orders_dataset.csv')
display(orders.head())
print("\n--- Data of orders ---")
orders.info()
```

```python
date_columns = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
    ]
for col in date_columns:
  orders[col] = pd.to_datetime(orders[col])
print("---Updated Data Types---")
print(orders.dtypes)
print('\n')
```

```python
print('Missing Values Count')
print(orders.isnull().sum())
print('\n')
```

```python
# Исправленный вариант удаления пропущенных значений из отфильтрованного датафрейма
orders_delivered = orders[orders['order_status'] == 'delivered'].copy()
orders_delivered.dropna(subset=['order_delivered_customer_date'], inplace=True)

print(f"Осталось доставленных заказов без пропусков: {len(orders_delivered)}")
```

```python
orders_delivered = orders[orders['order_status'] == 'delivered'].copy()
orders_delivered.dropna(subset=['order_delivered_customer_date'], inplace=True)
print(f"Original dataset size: {orders.shape[0]} rows")
print(f"Cleaned dataset size: {orders_delivered.shape[0]} rows")
```

```python
order_items = pd.read_csv(path + 'olist_order_items_dataset.csv')
sales_data = pd.merge(orders_delivered,order_items,on='order_id',how='inner')
```

```python
print(f'\nMerged dataset size:{sales_data.shape[0]} rows')
display(sales_data.head())
```
